Next before we proceed we need to merge our fraud data with the ip address to see which are having suspisious activity

```python
#convert ip address in fraud data and ip address in ip_address_data to float
Fraud_Data['ip_address']=Fraud_Data['ip_address'].astype(float)
IpAddress_to_Country['lower_bound_ip_address']=IpAddress_to_Country['lower_bound_ip_address'].astype(float)
IpAddress_to_Country['upper_bound_ip_address']=IpAddress_to_Country['upper_bound_ip_address'].astype(float)


```

```python
#merge them
def find_country(ip):
    row = IpAddress_to_Country[(IpAddress_to_Country['lower_bound_ip_address'] <= ip) & (IpAddress_to_Country['upper_bound_ip_address'] >= ip)]
    return row['country'].values[0] if not row.empty else "Unknown"

Fraud_Data['country'] = Fraud_Data['ip_address'].apply(find_country)
```

Now we can create a transaction frequency and velocity

```python
path_4='C:/Users/basleal/Desktop/tutorial/kaim-week-8-9/data/Fraud_Data_ip_merged.csv'
fraud_data_ip=pd.read_csv(path_4)
fraud_data_ip
```

```python
fraud_data_ip.info()
```

```python
# Convert timestamps to datetime format
fraud_data_ip['signup_time'] = pd.to_datetime(fraud_data_ip['signup_time'])
fraud_data_ip['purchase_time'] = pd.to_datetime(fraud_data_ip['purchase_time'])
```

```python
# Calculate transaction velocity (time difference between signup and purchase in hours)
fraud_data_ip['time_diff_hours'] = (fraud_data_ip['purchase_time'] - fraud_data_ip['signup_time']).dt.total_seconds() / 3600
```

```python
# Calculate transaction count per user (transaction frequency)
fraud_data_ip['transaction_count'] = fraud_data_ip.groupby('user_id')['user_id'].transform('count')
```
